```python
# === EXERCISE: Sample AI Project Data + Basic Stats ===

# 1. List of 10 daily hours spent on AI projects (simple list of numbers)
project_hours = [4.5, 6.0, 2.5, 8.0, 5.5, 3.0, 7.5, 1.5, 9.0, 4.0]

print("Raw project hours:", project_hours)
print("Total projects tracked:", len(project_hours))
```

```python
# 2. Dictionary version – more realistic (project name → hours)
ai_projects = {
    "Data Pipeline Optimization": 4.5,
    "Model Training Experiment": 6.0,
    "Prompt Engineering Research": 2.5,
    "Fine-tuning Llama-3": 8.0,
    "Data Cleaning Sprint": 5.5,
    "EDA on New Dataset": 3.0,
    "Build Streamlit Dashboard": 7.5,
    "Read Research Paper": 1.5,
    "Implement RAG System": 9.0,
    "Debug GPU Memory Issue": 4.0
}

print("AI Projects Dictionary")
for project, hours in ai_projects.items():
    print(f"  . {project}: {hours} hrs")
```

```python
# 3. Calculations – the stats every data script needs
total_hours = sum(project_hours) #total
average_hours = total_hours / len(project_hours) #mean
max_hours = max(project_hours) #maximum
max_project = max(ai_projects, key=ai_projects.get) # which project took long

print("=== BASIC STATS SUMMARY ===")
print(f"Total hours this week     : {total_hours} hrs")
print(f"Average hours per project : {average_hours} hrs")
print(f"Most time-intensive task  : {max_hours} hrs")
print(f"Productivity note         : You crushed {max_project} 🔥")

```
